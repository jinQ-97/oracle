```python
import sqlite3
# isolation_level : 트랜잭션
# autocommit 중지
conn = sqlite3.connect("./test.db",isolation_level=None)
# sql 을 데이터베이스에 전달하고 결과를 받아오는 객체
cusor = conn.cursor()
```

```python
# 테이블 생성
sql = """create table if not exists users(id interger primary key,
username text,
email text,
phone text,
website text,
regdate text
),
"""
cursor.execute(sql)
```

```python
from datetime import datetime
now = datetime.now()
NowDateTime = now.strftime("%y-%m-%d %h:%m:%s")
NowDateTime
```

```python
# insert 구문 : 튜플 형태로
sql ="insert into uesrs(id,username,email,phone,website,regdate) values(1,'kim','kim@naver.com','010-1234-5678','kim.com',?)"
cusor.execute(sql,(NowDateTime,))
```

```python
sql ="insert into uesrs(id,username,email,phone,website,regdate) values(?,?,?,?,?,?)"
cusor.execute(sql,(2,'park','park@naver.com','010-4321-9876','park.com',NowDateTime,))
```

```python
user_list = [
    (3,'hong','hong@naver.com','010-1234-5678','hong.com',NowDateTime)
    (4,'cho','cho@naver.com','010-1234-5678','cho.com',NowDateTime)
    (5,'yoo','yoo@naver.com','010-1234-5678','yoo.com',NowDateTime)
]
cusor.executemany(sql,user_list)
```
